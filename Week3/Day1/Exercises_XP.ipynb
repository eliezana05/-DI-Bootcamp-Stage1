```python
# Week 3 — Day 1: Intro to NumPy — Exercises XP
# Source: Octopus chapter 3377. Random outputs are reproducible examples.
import numpy as np
rng = np.random.default_rng(42)

# Exercise 1: integers from 0 to 9
numbers = np.arange(10)
print('1:', numbers)
np.testing.assert_array_equal(numbers, list(range(10)))

# Exercise 2: convert floats to integers (truncate towards zero)
integers = np.array([3.14, 2.17, 0, 1, 2]).astype(int)
print('2:', integers)
np.testing.assert_array_equal(integers, [3, 2, 0, 1, 2])

# Exercise 3: 3 by 3 matrix containing 1 to 9
matrix = np.arange(1, 10).reshape(3, 3)
print('3:\n', matrix)
assert matrix.shape == (3, 3) and matrix[2, 2] == 9

# Exercise 4: random 4 by 5 array
random_array = rng.random((4, 5))
print('4:\n', random_array)
assert random_array.shape == (4, 5)
assert np.all((random_array >= 0) & (random_array < 1))

# Exercise 5: second row, using zero-based indexing
array = np.array([[21,22,23,22,22], [20,21,22,23,24], [21,22,23,22,22]])
second_row = array[1, :]
print('5:', second_row)
np.testing.assert_array_equal(second_row, [20,21,22,23,24])

# Exercise 6: reverse the 1D array
reversed_array = numbers[::-1]
print('6:', reversed_array)
np.testing.assert_array_equal(reversed_array, [9,8,7,6,5,4,3,2,1,0])

# Exercise 7: 4 by 4 identity matrix
identity = np.eye(4)
print('7:\n', identity)
np.testing.assert_array_equal(identity @ np.arange(4), np.arange(4))
assert identity.sum() == 4

# Exercise 8: sum and average
array_sum, array_average = numbers.sum(), numbers.mean()
print('8: Sum:', array_sum, 'Average:', array_average)
assert array_sum == 45 and array_average == 4.5

# Exercise 9: reshape integers 1 to 20
reshaped = np.arange(1, 21).reshape(4, 5)
print('9:\n', reshaped)
assert reshaped.shape == (4, 5) and reshaped[-1, -1] == 20

# Exercise 10: Boolean selection of odd values
odd_numbers = numbers[numbers % 2 != 0]
print('10:', odd_numbers)
np.testing.assert_array_equal(odd_numbers, [1,3,5,7,9])
print('All 10 exercises verified successfully.')
```
