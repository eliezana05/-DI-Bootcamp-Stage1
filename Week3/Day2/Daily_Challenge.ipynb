```python
# Week 3 Day 2 Daily Challenge — Global Power Plant Database
# Source: Octopus chapter 3394, WRI database v1.3.0 supplied by the course.
# Installed capacity (MW) is not actual electricity generation (GWh).
import io, zipfile, requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from IPython.display import display, Markdown

url = 'https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%206%20-%20Applications%20for%20Data%20Analysis/W6D2%20-%20Advanced%20Numpy/globalpowerplantdatabasev130.zip'
response = requests.get(url, timeout=90)
response.raise_for_status()
archive = zipfile.ZipFile(io.BytesIO(response.content))
csv_names = [name for name in archive.namelist() if name.endswith('.csv') and '__MACOSX' not in name]
assert len(csv_names) == 1, csv_names
raw = pd.read_csv(archive.open(csv_names[0]), low_memory=False)
print('Raw dimensions:', raw.shape)
missing_before = raw.isna().sum().sort_values(ascending=False)
display(missing_before.rename('Missing values').to_frame())

# Clean IDs, categorical labels and numeric columns explicitly.
plants = raw.drop_duplicates(subset=['gppd_idnr']).copy()
for column in ['capacity_mw','latitude','longitude','commissioning_year']:
    plants[column] = pd.to_numeric(plants[column],errors='coerce').to_numpy(dtype=np.float64)
plants.loc[plants['capacity_mw'] < 0,'capacity_mw'] = np.nan
plants.loc[~plants['latitude'].between(-90,90),'latitude'] = np.nan
plants.loc[~plants['longitude'].between(-180,180),'longitude'] = np.nan
plants.loc[~plants['commissioning_year'].between(1800,2026),'commissioning_year'] = np.nan
for column in ['country_long','primary_fuel']:
    plants[column] = plants[column].fillna('Unknown')
# Do not fabricate commissioning years, locations or generation by imputing means.
# Exclude missing capacity only from capacity analyses, not from plant counts.
capacity_data = plants.dropna(subset=['capacity_mw']).copy()
print('Clean rows:',len(plants),'Capacity-analysis rows:',len(capacity_data))
num_cols = plants.select_dtypes(include=np.number).columns
display(plants[num_cols].agg(['mean','median','std']).T)
country_counts = plants['country_long'].value_counts()
fuel_counts = plants['primary_fuel'].value_counts()
display(country_counts.rename('Plant count').head(15).to_frame(),fuel_counts.rename('Plant count').to_frame())

# NumPy statistics on installed capacity by fuel.
fuel_records=[]
for fuel, group in capacity_data.groupby('primary_fuel'):
    values=group['capacity_mw'].to_numpy()
    fuel_records.append([fuel,len(values),np.mean(values),np.median(values),np.std(values,ddof=1) if len(values)>1 else np.nan,np.sum(values)])
fuel_stats=pd.DataFrame(fuel_records,columns=['Fuel','Count','Mean MW','Median MW','Std MW','Total MW']).set_index('Fuel')
display(fuel_stats.sort_values('Total MW',ascending=False).round(2))

# Pre-specified comparison: coal vs gas, H0 equal mean installed capacity.
# Welch's test permits unequal variances and sample sizes; alpha=0.05.
coal=capacity_data.loc[capacity_data.primary_fuel=='Coal','capacity_mw'].to_numpy()
gas=capacity_data.loc[capacity_data.primary_fuel=='Gas','capacity_mw'].to_numpy()
welch=stats.ttest_ind(coal,gas,equal_var=False)
mean_difference=coal.mean()-gas.mean()
se=np.sqrt(coal.var(ddof=1)/len(coal)+gas.var(ddof=1)/len(gas))
dof=(coal.var(ddof=1)/len(coal)+gas.var(ddof=1)/len(gas))**2 / ((coal.var(ddof=1)/len(coal))**2/(len(coal)-1)+(gas.var(ddof=1)/len(gas))**2/(len(gas)-1))
ci=mean_difference+np.array([-1,1])*stats.t.ppf(0.975,dof)*se
print('Welch coal vs gas:',welch,'Mean difference MW:',mean_difference,'95% CI:',ci)
print('Reject equal means at alpha=.05' if welch.pvalue<.05 else 'Do not reject equal means at alpha=.05')
print('Observational database, not a random global sample. Coverage, skew, country mix and outliers limit inference; fuel does not establish causation.')

# Commissioning-year trends: counts and capacity in database by commissioning decade.
dated=capacity_data.dropna(subset=['commissioning_year']).copy()
dated['Decade']=(np.floor(dated.commissioning_year.to_numpy()/10)*10).astype(int)
decade_mix=pd.crosstab(dated['Decade'],dated['primary_fuel']).sort_index()
decade_shares=decade_mix.div(decade_mix.sum(axis=1),axis=0)
print('Rows with known commissioning year:',len(dated),'of',len(capacity_data))
display(decade_mix.tail(10))
fig,axes=plt.subplots(2,2,figsize=(16,11),constrained_layout=True)
country_counts.head(10).sort_values().plot.barh(ax=axes[0,0]); axes[0,0].set(title='Countries with most recorded plants',xlabel='Plant count')
fuel_stats['Total MW'].sort_values().plot.barh(ax=axes[0,1]); axes[0,1].set(title='Installed capacity by primary fuel',xlabel='MW')
sns.histplot(capacity_data['capacity_mw'],bins=60,ax=axes[1,0]); axes[1,0].set(title='Capacity distribution',xlabel='Installed capacity (MW)'); axes[1,0].set_yscale('log')
decade_shares.loc[decade_shares.index>=1900].plot.area(ax=axes[1,1]); axes[1,1].set(title='Fuel share of plants by commissioning decade',xlabel='Commissioning decade',ylabel='Share of recorded plants'); axes[1,1].legend(fontsize=7,ncol=2)
plt.show()
geo=capacity_data.dropna(subset=['latitude','longitude'])
fig,ax=plt.subplots(figsize=(13,6))
points=ax.scatter(geo.longitude,geo.latitude,c=np.log1p(geo.capacity_mw.to_numpy()),s=5,alpha=.45,cmap='viridis')
ax.set(title='Recorded plant locations (longitude/latitude)',xlabel='Longitude',ylabel='Latitude',xlim=(-180,180),ylim=(-90,90)); fig.colorbar(points,ax=ax,label='log(1 + capacity MW)'); plt.show()

# Matrix operations: correlation/PCA of capacity, location and fuel indicators.
top_fuels=fuel_counts.head(4).index
features=geo[['capacity_mw','latitude','longitude']].copy()
features['capacity_mw']=np.log1p(features.capacity_mw)
for fuel in top_fuels:
    features['fuel_'+fuel]=(geo.primary_fuel==fuel).astype(float)
X=features.to_numpy(dtype=float); std=X.std(axis=0,ddof=1)
assert np.all(std>0)
Z=(X-X.mean(axis=0))/std
correlation=Z.T @ Z/(len(Z)-1)
eigenvalues,eigenvectors=np.linalg.eigh(correlation)
order=np.argsort(eigenvalues)[::-1]; eigenvalues=eigenvalues[order]; eigenvectors=eigenvectors[:,order]
variance_ratio=eigenvalues/eigenvalues.sum()
display(pd.DataFrame(correlation,index=features.columns,columns=features.columns).round(3))
print('Eigenvalues:',eigenvalues,'Explained variance:',variance_ratio)
np.testing.assert_allclose(correlation @ eigenvectors,eigenvectors*eigenvalues,atol=1e-9)
np.testing.assert_allclose(np.diag(correlation),1,atol=1e-9)
# Fuel labels are never treated as ordinal numbers. One-hot features allow
# examining associations, but PCA loadings depend on coding and standardization.
# Eigenvectors identify directions of variation, not causes or policy targets.
mask=np.logical_and(capacity_data.capacity_mw.to_numpy()>=1000,capacity_data.primary_fuel.isin(['Coal','Gas']).to_numpy())
print('NumPy mask: coal/gas plants with at least 1000 MW:',np.count_nonzero(mask))
display(capacity_data.loc[mask,['name','country_long','primary_fuel','capacity_mw']].head(10))

leader=fuel_stats['Total MW'].idxmax()
report=f'''# Global power plant analysis — findings
- Source: WRI Global Power Plant Database v1.3.0, course-provided archive.
- Records analyzed: {len(plants):,}; capacity records: {len(capacity_data):,}.
- Most recorded plants: {country_counts.idxmax()} ({country_counts.max():,}).
- Most recorded plants by fuel: {fuel_counts.idxmax()} ({fuel_counts.max():,}).
- Highest total installed capacity: {leader} ({fuel_stats.loc[leader,'Total MW']:,.0f} MW).
- Coal mean capacity: {coal.mean():.2f} MW; gas mean capacity: {gas.mean():.2f} MW. Difference: {mean_difference:.2f} MW (Welch p={welch.pvalue:.3g}; 95% CI {ci[0]:.2f} to {ci[1]:.2f} MW).
- Known commissioning years: {len(dated):,}/{len(capacity_data):,}. Decade charts describe recorded plants still in this snapshot, not historical national energy mixes; missing dates and retired plants limit conclusions.
- NumPy handles numeric conversion, grouped statistics, masks, matrix multiplication and eigendecomposition. Pandas organizes cleaning and groups; Matplotlib and Seaborn display distributions, locations and time trends.
- Capacity is potential power (MW), not actual output/annual generation (GWh). Missing actual generation is not replaced with capacity. This snapshot is incomplete and observational.
- PCA first component explains {variance_ratio[0]:.1%} of standardized feature variance; geographic and fuel coding choices affect this result.
'''
display(Markdown(report))
# Keep a standalone report for the README deliverable.
with open('README.md','w') as handle:
    handle.write(report)
assert plants.gppd_idnr.is_unique
assert np.isfinite(capacity_data.capacity_mw).all()
assert np.all(capacity_data.capacity_mw>=0)
print('Power plant analysis checks passed.')
```
